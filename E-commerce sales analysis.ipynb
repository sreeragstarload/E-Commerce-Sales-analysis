# 4.1 create a working copy

```python
df_clean = df.copy()
```

# 4.2 standardize column Names

```python
df_clean.columns = (
    df_clean.columns
    .str.strip()
    .str.lower()
    .str.replace(r'\s+', '_', regex=True)
)

print(df_clean.columns.tolist())
```

# 4.3 check missing values

```python
df_clean.isnull().sum()
```

# 4.4 Handle missing values

```python
# NUMERICAl columns
num_cols = df_clean.select_dtypes(
    include=np.number
).columns

for col in num_cols:
    if df_clean[col].isnull().sum() > 0:
        df_clean[col] = df_clean[col].fillna(
            df_clean[col].median()
        )
#categorical columns
cat_cols = df_clean.select_dtypes(
    include=['object', 'category']
).columns

for col in cat_cols:
    if df_clean[col].isnull().sum() > 0:
        df_clean[col] =df_clean[col].fillna('Unknown')
```

# 4.5 Check Duplicate records

```python
df_clean.duplicated().sum()
```

# 4.6 Handle Duplicates

```python
df_clean = df_clean.drop_duplicates()
print("Remaining Duplicate:",
      df_clean.duplicated().sum())
```

# 4.7 check data types

```python
df_clean.dtypes
```

```python
numeric_candidates = [
    'quantity',
    'unit_price',
    'discount',
    'sales',
    'profit'
]

for col in numeric_candidates:
    if col in df_clean.columns:
        df_clean[col] = pd.to_numeric(
            df_clean[col],
            errors='coerce'
        )
```

# 4.8 Convert & Extract Time features

```python
# Convert order_date to datetime

df_clean['order_date'] = pd.to_datetime(df_clean['order_date'], errors='coerce')

# Extract Month and Year for trend analysis

df_clean['order_month'] = df_clean['order_date'].dt.to_period('M').astype(str)

print(df_clean[['order_date', 'order_month']].head())
```

```python
df_clean.isnull().sum()
```

# 4.9 Handle invalid values

```python
for col in['quantity','unit_price', 'sales']:
    if col in df_clean.columns:
        print("\n",col)
        print(df_clean[df_clean[col] < 0])
```

# 4.10checck outliers

```python
for col in df_clean.select_dtypes(
    include=np.number
).columns:

    Q1 = df_clean[col].quantile(0.25)
    Q3 = df_clean[col].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = df_clean[
        (df_clean[col] < lower) |
        (df_clean[col] > upper)
    ]

    print(col, "Outliers:", len(outliers))
```

# 5.1 create primary key

```python
df_db = df_clean.dropna(
    subset=['product_id']
).copy()

df_db['product_id'] = (
    df_db['product_id'].astype(str)
)

df_db = df_db.reset_index(drop=True)

df_db.insert(
    0,
    'record_id',
    range(1, len(df_db) + 1)
)

df_db.head()
```

*record_id is each sales record to unique identifier*

# 5.2 identify Foreign key

```python
products = df_db[['product_id']].drop_duplicates()

products = products.reset_index(drop=True)

print("Product IDs:", len(products))
print("Unique Product IDs:",
      products['product_id'].is_unique)
```

# 5.3 Create related tables

```python
sales_table = df_db.copy()

products_table = products.copy()

print("Products Table:")
display(products_table.head())

print("Sales Table:")
display(sales_table.head())
```

# 5.4 eastablish tables relationships

```python
import sqlite3

conn = sqlite3.connect('ecommerce_sales.db')

conn.execute("PRAGMA foreign_keys = ON")

cursor = conn.cursor()

cursor.execute("DROP TABLE IF EXISTS Sales")
cursor.execute("DROP TABLE IF EXISTS Products")

cursor.execute("""
CREATE TABLE Products (
    product_id TEXT PRIMARY KEY
)
""")

# Build Sales table using actual dataset columns
definitions = [
    '"record_id" INTEGER PRIMARY KEY',
    '"product_id" TEXT NOT NULL'
]

for col in sales_table.columns:
    if col in ['record_id', 'product_id']:
        continue

    if pd.api.types.is_numeric_dtype(sales_table[col]):
        sql_type = 'REAL'
    else:
        sql_type = 'TEXT'

    definitions.append(f'"{col}" {sql_type}')

definitions.append(
    'FOREIGN KEY (product_id) REFERENCES Products(product_id)'
)

cursor.execute(
    'CREATE TABLE Sales (' + ', '.join(definitions) + ')'
)

products_table.to_sql(
    'Products', conn, if_exists='append', index=False
)

sales_table.to_sql(
    'Sales', conn, if_exists='append', index=False
)

conn.commit()

print("Database tables created successfully.")
```

# verify the relationship

```python
pd.read_sql_query(
    "PRAGMA foreign_key_check",
    conn
)
```

# 6.1 calculate key performance indicators(KPLS)

```python
total_sales = df_clean['sales'].sum()
total_profit = df_clean['profit'].sum()
total_quantity = df_clean['quantity'].sum()

average_sales = df_clean['sales'].mean()
average_profit = df_clean['profit'].mean()

profit_margin = (
    (total_profit / total_sales) * 100
    if total_sales != 0 else 0
)

print("Total Sales:", total_sales)
print("Total Profit:", total_profit)
print("Total Quantity:", total_quantity)
print("Average Sales:", average_sales)
print("Average Profit:", average_profit)
print("Profit Margin:", profit_margin)
```

# 6.2 Sales analysis

```python
print("Total Sales:", df_clean['sales'].sum())
print("Average Sales:", df_clean['sales'].mean())
print("Maximum Sales:", df_clean['sales'].max())
print("Minimum Sales:", df_clean['sales'].min())
```

# 6.3 Profit analysis

```python
print("Total Profit:", df_clean['profit'].sum())
print("Average Profit:", df_clean['profit'].mean())
print("Maximum Profit:", df_clean['profit'].max())
print("Minimum Profit:", df_clean['profit'].min())
```

# 6.4 quantity analysis

```python
print("Total Quantity:", df_clean['quantity'].sum())
print("Average Quantity:", df_clean['quantity'].mean())
print("Maximum Quantity:", df_clean['quantity'].max())
print("Minimum Quantity:", df_clean['quantity'].min())
```

```python
if 'discount' in df_clean.columns:

    print("Average Discount:",
          df_clean['discount'].mean())

    print("Maximum Discount:",
          df_clean['discount'].max())

    discount_analysis = df_clean.groupby(
        'discount'
    ).agg(
        total_sales=('sales', 'sum'),
        total_profit=('profit', 'sum')
    )

    display(discount_analysis)
```

# 6.6 product analysis

```python
product_analysis = df_clean.groupby(
    'product_id'
).agg(
    total_sales=('sales', 'sum'),
    total_profit=('profit', 'sum'),
    total_quantity=('quantity', 'sum')
).sort_values(
    by='total_sales',
    ascending=False
)

display(product_analysis)
```

# 6.7 payment method analysis

```python
payment_analysis = df_clean.groupby(
    'payment_method'
).agg(
    total_sales=('sales', 'sum'),
    total_profit=('profit', 'sum')
).sort_values(
    by='total_sales',
    ascending=False
)

display(payment_analysis)
```

# 6.8 shipping mode analysis

```python
shipping_analysis = df_clean.groupby(
    'ship_mode'
).agg(
    total_sales=('sales', 'sum'),
    total_profit=('profit', 'sum')
).sort_values(
    by='total_sales',
    ascending=False
)

display(shipping_analysis)
```

# 6.9 order status analysis

```python
# 6.10 Monthly Sales Trend
monthly_trend = df_clean.groupby('order_month').agg(
    total_sales=('sales', 'sum'),
    total_profit=('profit', 'sum'),
    order_count=('order_id', 'count')
).sort_index()

display(monthly_trend)
```
